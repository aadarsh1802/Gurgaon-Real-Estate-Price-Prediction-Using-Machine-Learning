```python
import pandas as pd
import numpy as np
import seaborn as sns
```

```python
df = pd.read_csv('../data/raw/gurgaon_property.csv')
```

```python
df.head()
```

```python
df['area_type'] = df['areaWithType'].str.extract(
    r'\)\s*(.*)'
)[0]
```

```python
df['area_type'].value_counts()
```

```python
df['area_sqm'] = df['areaWithType'].str.extract(r'\(([\d.]+)\s*sq\.m\.\)')
```

```python
df['build_up'] = df['area_sqm'].astype(float) * 10.7639
```

```python
df['area_unit'] = df['areaWithType'].str.extract(
    r'([\d.]+)\s*(sq\.ft\.|sq\.m\.|sq\.yd\.)',
    expand=False
)[1]
```

```python
df['area_unit'].value_counts()
```

```python
def convert_to_sqft(row):
    value = row['area_value']
    unit = row['area_unit']
    
    if pd.isna(value) or pd.isna(unit):
        return np.nan
    
    if unit == 'sq.ft.':
        return value
    
    elif unit == 'sq.m.':
        return value * 10.7639
    
    elif unit == 'sq.yd.':
        return value * 9
    
    return np.nan
```
